## Abstract

This project examines how anime production trends changed from 1990 to 2025, with a focus on two influential genres: Shounen and Slice of Life. We were interested in whether Slice of Life experienced a modern boom, whether Shounen began to slow down, and whether Slice of Life ever overtook Shounen in production. To study this, we used a cleaned anime production dataset derived from MyAnimeList, created binary indicators for Shounen and Slice of Life titles, removed titles tagged as both genres, converted release dates into yearly measures, and filtered out very short works. We then compared yearly production counts, genre market share relative to all anime released in a year, 5-year moving-average trends, genre overlap patterns, and regression-based growth trajectories.

Our results show that Shounen remained stronger than Slice of Life in absolute production and market share throughout 1990–2025. However, the two genres followed different trajectories: Shounen generally plateaued and began to lose momentum in the mid-to-late 2000s, while Slice of Life grew quickly starting around 2008–2010. A regression on relative market-share growth supported this pattern, showing that Slice of Life’s growth trend increased significantly faster than Shounen’s over time. At the same time, our genre co-occurrence analysis suggests that Shounen benefits from stronger overlap with already dominant genres such as Action, Comedy, and Adventure, while Slice of Life is comparatively more isolated. Overall, our findings suggest that Slice of Life has not yet surpassed Shounen in raw production or market share, but it has become one of the fastest-growing genre trends in recent anime history and may continue narrowing the gap.

## Background and Prior Work

Popular media is a reflection of the interests of the majority and changes over time and place through the influence of culture and environment. Trends can arise when certain genres attract a significant following and creators appeal to this interest, leading to increased production in media with common themes. In terms of anime, two genres that have had a consistently large following and interest are shonen and slice-of-life. Many iconic anime fall under the shonen genre such as Naruto and Dragon Ball and are usually characterized by action and fantasy themes. Similarly, figures from slice-of-life anime such as Lucky Star and Love Live have become easily recognizable from their long-standing popularity and are characterized by lighthearted and relatable stories. With shonen and slice-of-life being popular genres in Japanese media for a long time, we wanted to better understand any potential changes in production between anime of these two genres over the years.

Many shonen series first find their start by having a chapter published in the Weekly Shonen Jump magazine before being compiled into a full manga. The target audience of this magazine is usually teen and young adult boys, which is seen through the frequent use of male protagonists who reflect this audience. The article “The Rise and Fall of Weekly Shonen Jump: A Look at the Circulation of Weekly Jump” focuses on the amount of sales of the Weekly Shonen Jump magazine from 1970 to 2005 and takes note of its peak popularity and when it begins to decline <a href="#ref1">[1]</a>. Observing sales of the magazine, it began its height of popularity in 1993 but “…between 1995 and 2005, while other magazines' sales stayed fairly constant, Shonen Jump's circulation decreased dramatically” <a href="#ref1">[1]</a>. At the same time, competing magazines like Weekly Shonen Magazine also found a slight decline in sales around a similar time period <a href="#ref1">[1]</a>. Comparing this change in sales to the titles in the editions of this time, the decrease in sales coincides with the ending of popular series like Dragon Ball <a href="#ref1">[1]</a>. This study was done in 2007 and only observes sales up to 2005. Additionally, the data solely focuses on the popularity of titles being published through this particular magazine, which may not reflect titles in other magazines or other platforms like television.

In “Trends of Fiction in 2000s Japanese Pop Culture” published in the Electronic Journal of Contemporary Japanese Studies, Motoko Tanaka studies the fall of “the fictional age” (kyokō no jidai) and the beginning of “the age of impossibility” zeitgeists in Japan, along with how that change correlated with the rise and fall of certain anime genres popular among teen-young adult Japanese men <a href="#ref2">[2]</a>. Tanaka argues that the 1970–1995 “fictional age,” which was heralded in by the failure of the Anpo and Zenkyōtō student activism movements, was defined by the desire of young Japanese men to “…seek ideals not in political movements and changes in the real world, but rather in imagined settings such as theme parks and in fictional characters” <a href="#ref2">[2]</a>. While the article doesn’t directly mention shonen anime being popular, it can be logically inferred that anime with strong, morally just male protagonists in fantasy settings fit this cultural desire. Tanaka then discusses the fall of the “fictional age” and the beginning of "the age of impossibility,” influenced by catastrophes such as the 1995 Kobe Earthquake, the Aum sarin gas attack, and 9/11, as well as Japan’s ‘lost decade’ <a href="#ref2">[2]</a>. During this period, more optimistic genres such as traditional shonen saw a decline in prominence, while darker genres such as Sekaikei and Sabaibukei (death game narratives) gained attention <a href="#ref2">[2]</a>. Tanaka also notes that, during the late 2000s, Nichijōkei (slice-of-life) anime began to surge in popularity <a href="#ref2">[2]</a>. Although the paper focuses more on cultural analysis than statistical evidence and is limited mainly to the 1990s–2000s, it still identifies clear genre trends and their cultural context, which helped inform our hypothesis.

#### Section 1A - Number of anime produced per year (Slice of Life vs Shounen)

In this line graph, the start years from 1990-2025 is plotted on the x-axis which describes the year a title is first aired. On the y-axis, it displays the number of anime that were produced in a given year. Overall within a given year, shonen titles tend to be produced a lot more than slice-of-life series. Shonen titles also began to increase in production much more rapidly throughout the years than slice-of-life especially from 2000-2010 where shonen increased from around 20 titles per year to nearly 50 titles per year within that time frame while slice-of-life steadily maintained around less than 10 titles every year. However, by 2010, both genres share similar shapes in increasing trends but slice-of-life production is still significantly less than shonen.

```python
#creates a dataframe with total Shounen and Slice of Life produced per year respectively
cleaned_data = df.groupby('start_year')[['is_shounen','is_slice_of_life']].sum().reset_index()

#filters out anime from before 1990 and after 2025
cleaned_data = cleaned_data[(cleaned_data['start_year'] >= 1990) & (cleaned_data['start_year'] <= 2025)]

#plot generation 
sns.lineplot(data = cleaned_data,x='start_year',y='is_shounen',label = 'Shounen')
sns.lineplot(data = cleaned_data,x='start_year',y='is_slice_of_life',label = 'Slice of Life')
plt.ylabel("Number anime produced")
plt.title('Anime Produced per year (Shounen/Slice of Life)')
```

#### Section 1B - Genre Market Share (Slice of Life vs Shounen)

Similarly, this graph displays the start years from 1990-2025 on the x-axis. However, this time, the y-axis displays market shares in percentages which describe the proportions of titles produced within a year that were shonen or slice-of-life. Under shonen from 1990-2010, market share proportions regularly stay above 10% where every few years market shares rapidly spike up to 20% or more before once again falling rapidly back to around 10%. By 2010, the change in trends showed a significant decline in market shares. On the other hand, for slice-of-life, the genre maintains a low market share of under 5% from 1990-2010 and begins a steady increase in market share after 2010 but still remains much lower than shonen. 

```python

#filters out anime from before 1990 and after 2025
df = df[(df['start_year'] >= 1990) & (df['start_year'] <= 2025)]

#sum total Slice of Life,Shounen and total anime per year respectively
two_genre_yearly = df.groupby('start_year')[['is_shounen','is_slice_of_life']].sum()
all_genre_yearly = df.groupby('start_year').size()

#creates proportions for chosen genres vs all genres per year
two_genre_proportions = two_genre_yearly.div(all_genre_yearly,axis = 0)

#plot generation 
sns.lineplot(data = two_genre_proportions,x='start_year',y='is_shounen',label = 'Shounen')
sns.lineplot(data = two_genre_proportions,x='start_year',y='is_slice_of_life',label = 'Slice of Life')
plt.ylabel("market share(%)")
plt.title('Genre Market Share (Slice of Life vs Shounen)')
```

#### Section 1C - Smoothed Genre Market Share (Slice of Life vs Shounen)

Since the yearly genre frequency flutuates, the general trend is not very easy to see. Thus, we apply a 5 years moving average (MA) to the market share of each genre. From this graph, we can easily see that the market share of Shounen starts to decrease at around 2010 while the market share of Slice of Life starts to climb at around 2007.

```python
window_size = 5
smoothed_proportions = two_genre_proportions.rolling(window=window_size).mean().dropna()

sns.lineplot(data=smoothed_proportions, x='start_year', y='is_shounen', label='Shounen (5-yr MA)')
sns.lineplot(data=smoothed_proportions, x='start_year', y='is_slice_of_life', label='Slice of Life (5-yr MA)')

plt.ylabel("Market Share")
plt.title('Genre Market Share - 5-Year Moving Average (Slice of Life vs Shounen)')
plt.show()
```

#### Section 2B - Anime Produced per Year by Genre (Top 5 + Shounen + Slice of Life + Other)

Similar to the previous graph, this graph utilizes the same axes but displays the data in a histogram where anime produced by year are split into the top 5 most popular genres aside from our two genres of focus, shonen and slice-of-life. Aside from our two genres of focus, the top five genres of anime in this graph include comedy, action, fantasy, adventure, and sci-fi. Overall, the majority of anime produced in a year over the decades are categorized as either action or comedy with comedy being the biggest genre. In the recent decade, compared to other top 5 genres, slice-of-life usually takes up a larger portion of anime produced in a year than adventure and shares a similar portion with sci-fi. Meanwhile, shonen takes up a larger portion than sci-fi, adventure, and slice-of-life while it shares a similar portion with fantasy. 

```python
df = pd.read_csv("data/02-processed/preprocessed_anime_data.csv")

# Exclude 2026+
df = df.dropna(subset=["start_year"]).copy()
df["start_year"] = df["start_year"].astype(int)
df = df[df["start_year"] < 2026].copy()

# top 5 genres overall (exclude Slice of Life + Shounen since they get their own bins)
all_genres = (
    df["genres"].dropna()
      .str.split(";")
      .explode()
      .str.strip()
      .replace("", np.nan)
      .dropna()
)
top5 = (
    all_genres[~all_genres.isin(["Slice of Life", "Shounen"])]
    .value_counts()
    .head(5)
    .index
    .tolist()
)

# --- assign each anime to exactly one bucket (priority: Shounen, Slice of Life, top5, Other) ---
def bucket_genre(row):
    if row.get("is_shounen", False):
        return "Shounen"
    if row.get("is_slice_of_life", False):
        return "Slice of Life"

    g = row.get("genres", None)
    if pd.isna(g):
        return "Other"
    gset = {x.strip() for x in str(g).split(";") if x.strip()}
    for genre in top5:
        if genre in gset:
            return genre
    return "Other"

df["genre_bucket"] = df.apply(bucket_genre, axis=1)

counts = (
    df.groupby(["start_year", "genre_bucket"])
      .size()
      .unstack(fill_value=0)
      .sort_index()
)

order = top5 + ["Shounen", "Slice of Life", "Other"]
counts = counts.reindex(columns=order, fill_value=0)

# --- plot ---
fig, ax = plt.subplots(figsize=(14, 6))
counts.plot(kind="bar", stacked=True, ax=ax, width=0.9)

ax.set_title("Anime Produced per Year by Genre (Top 5 + Shounen + Slice of Life + Other)")
ax.set_xlabel("Start Year")
ax.set_ylabel("Number of Anime")
ax.grid(axis="y", alpha=0.3)

years = counts.index.to_numpy()
tick_every = max(1, len(years) // 20)
ax.set_xticks(np.arange(0, len(years), tick_every))
ax.set_xticklabels(years[::tick_every], rotation=45, ha="right")

ax.legend(title="Bucket", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

print("Top 5 genres used:", top5)
```

#### Section 3C - How are genres related?
Knowing that most anime have multiple genres, we will use a heatmap to analyze if any genres of interest have any correlation to each other. We will take the top 20 genres including our target focus (Shounen and Slice of Life) to observe how frequently other genres appear in tandem to each other.
The heatmap provides a lot of interesting insights overall, like how martial arts is 100% correlated with action. However, we only want to focus on our two rows (Shounen and Slice of Life) for this analysis. Looking at the rows for Shounen, the % of genres that are also associated with it most frequently are Comedy, Action, and Adventure. For Slice of Life, it has a small correlation with the School genre. From this, we can conclude that our two focus genres do not have much genre overlap, as they have a 0% correlation with each other and their most frequent associated genres are different. Shounen does have a small correlation with the School genre, but as the School genre largely just consists of works with a setting in a school compared to a particular format of show, it is not of much interest. Going back to our previous chart analyzing the top 15 most frequent genres, we can draw some more connections from that chart given what is shown here.


```python
# one-hot genres
df_exploded = df['genres'].str.split(';').explode()
df_dummies = pd.get_dummies(df_exploded).groupby(level=0).max().astype(int)

# pick top 20 + slice of life
genres_to_plot = df_dummies.sum().nlargest(20).index.tolist()
if 'Slice of Life' in df_dummies.columns and 'Slice of Life' not in genres_to_plot:
    genres_to_plot.append('Slice of Life')
df_filtered = df_dummies[genres_to_plot]

# get co-ocurrance matrix
co_matrix_raw = df_filtered.T.dot(df_filtered)

# normalize by number of shows/genre
genre_totals = np.diag(co_matrix_raw)
co_matrix_norm = co_matrix_raw.div(genre_totals, axis=0)

# remove diag values
np.fill_diagonal(co_matrix_norm.values, 0)

plt.figure(figsize=(14, 11))
sns.heatmap(co_matrix_norm, annot=True, cmap='Blues', fmt='.0%', vmin=0, vmax=1)
plt.title('Normalized Genre Co-occurrence (Given Row Genre, % Chance of Column Genre)')
plt.ylabel('Given this genre')
plt.xlabel('What % are also this genre?')
plt.show()
```

In this graph, we can clearly see that the market share of Slice of Life skyrocks starting from around 2008. On the other hand, the market share of Shounen mostly plateau or even decreased from around 2010.

```python
# 1990 baseline value
baseline = smoothed_proportions.iloc[0]

# Calculate the percentage change relative to 1990
# ((Current Year - Baseline Year) / Baseline Year) * 100
growth_df = smoothed_proportions.copy()
relative_growth = ((growth_df - baseline) / baseline) * 100

sns.lineplot(data=relative_growth, x='start_year', y='is_shounen', label='Shounen Growth', marker='o')
sns.lineplot(data=relative_growth, x='start_year', y='is_slice_of_life', label='Slice of Life Growth', marker='o')

# baseline at 0%
plt.axhline(0, color='black', linestyle='--', alpha=0.5)

plt.ylabel("Growth (%) with respect to 1990")
plt.title('Genre Market Share Growth (1990 Baseline)')
plt.show()
```

To check if the market share growth of Slice of Life has outpaced that of Shounen, we run an OLS linear regression. We define the model as 

$Growth = \beta_0 + \beta_1 \cdot Year + \beta_2 \cdot Genre + \beta_3 \cdot (Year \times Genre)$

Note that $Genre$ is 0 for Shounen and 1 for Slice of Life as Shounene is alphabetically smaller than Slice of Life. Thus, Shounen serve as the baseline with the model being

$Growth = \beta_0 + \beta_1 \cdot Year + \beta_2 \cdot 0 + \beta_3 \cdot (Year \times 0) = \beta_0 + \beta_1 \cdot Year$

On the other hand, the model for Slice of Life is

$Growth = \beta_0 + \beta_1 \cdot Year + \beta_2 \cdot 1 + \beta_3 \cdot (Year \times 1) = \beta_0 + \beta_1 \cdot Year + \beta_2 + \beta_3 \cdot Year = (\beta_0 + \beta_2) + (\beta_1 + \beta_3) \cdot Year$

The variable $\beta_3$ indicates the difference in growth between the genres. 

```python
model_df = relative_growth.reset_index()
# from wide to long
model_df = pd.melt(model_df, 
                   id_vars=['start_year'], 
                   value_vars=['is_shounen', 'is_slice_of_life'],
                   var_name='Genre', 
                   value_name='Growth')
model_df['Genre'] = model_df['Genre'].replace({'is_shounen': 'Shounen', 'is_slice_of_life': 'Slice of Life'})
# set 1995 as the 0th year
model_df['Year_Index'] = model_df['start_year'] - 1995

# ols regression
model = smf.ols('Growth ~ Year_Index * Genre', data=model_df).fit()
model.summary()
```

Before diving into the analysis, we first verify if the fitted line is statistically robust. The JB value is 0.591, so we cannot reject the null which is the residuals are not normally distributed. However, the Durbin-Watson is 0.202 which indicates a positive autocorrelation in growth (what happens in 2010 is heavily dependent on what happened in 2009). This makes our model over confident and underestimate the standard error and p-value. Thus, we apply Heteroskedasticity and Autocorrelation Consistent errors (HAC) to address this issue.

```python
model_robust = smf.ols('Growth ~ Year_Index * Genre', data=model_df).fit(cov_type='HAC', cov_kwds={'maxlags': 1})
model_robust.summary()
```

From the result, we see that Adj. R-squared being 0.777 which is means that model fits the data decently well. $\beta_3$ which is Year_Index:Genre[T.Slice of Life] in the table. The fitted value is 8.0866 which means Slife of life grows 8.0866% more than Shounen on average yearly in the fitted line. The p-value is 0.000, which indicates a strong rejection of the null hypothesis.

To see this visually, we plot the fitted line with the actual data. We can see that the growth of Shounen is slowly going down while the growth of Slice of Life is trending up.

```python
model_df['Predicted_Growth'] = model_robust.predict(model_df)

# plot the actual data
sns.lineplot(data=model_df[model_df['Genre'] == 'Shounen'], 
             x='start_year', y='Growth', alpha=0.4, label='Shounen (Actual)')
sns.lineplot(data=model_df[model_df['Genre'] == 'Slice of Life'], 
             x='start_year', y='Growth', alpha=0.4, label='Slice of Life (Actual)')

# plot the ols prediction
sns.lineplot(data=model_df[model_df['Genre'] == 'Shounen'], 
             x='start_year', y='Predicted_Growth', color='tab:blue', label='Shounen (OLS Trend)')
sns.lineplot(data=model_df[model_df['Genre'] == 'Slice of Life'], 
             x='start_year', y='Predicted_Growth', color='tab:orange', label='Slice of Life (OLS Trend)')

# 0% line
plt.axhline(0, color='black', linestyle='--', alpha=0.5)

plt.ylabel("Cumulative Growth (%) since 1995")
plt.xlabel("Release Year")
plt.title('Genre Market Share Growth with OLS Trendlines (1990-2025)')
plt.legend(loc='upper left')

plt.show()
```
